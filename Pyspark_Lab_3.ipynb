### Union & UnionByName

```python
data1 = [(1, 'Adithya'), (2, 'Sam'), (3, 'Akriti')]
schema = "id string,name string"
df1 = spark.createDataFrame(data1, schema)
df1.display()
```

```python
data2 = [(4, 'Neha'), (5, 'Rohan'), (6, 'Joshi')]
schema = "id string,name string"
df2 = spark.createDataFrame(data2, schema)
df2.display()
```

```python
df1.union(df2).display()
```

```python
data3 = [('Neha', 4), ('Rohan', 5), ('Joshi', 6)]
schema = "name string,id string"
df3 = spark.createDataFrame(data3, schema)
df3.display()
```

```python
df1.union(df3).display()
```

```python
df1.unionByName(df3).display()
```

### String Functions

```python
df1.select(upper('name')).display()
```

```python
df.select(initcap('Item_Type').alias('Upper')).display()
```

### Date Functions
- current_date()
- date_add()
- date_sub()

```python
df.withColumn("Current_Date", current_date()).display()
```

```python
df.withColumn("Current_Date", current_timestamp()).display()
```

```python
df.withColumn("Future_Date", date_add(col('Current_Date'), 7)).display()
```

```python
df.withColumn("Past_Date", date_sub(col('Current_Date'), 7)).display()
```

```python
df.withColumn("Current_Date", current_date()).withColumn("Future_Date", date_add(col('Current_Date'), 7)).withColumn("Date_Diff", date_diff("Future_Date", "Current_Date")).display()
```

```python
df.withColumn("Formatted_Date", date_format(current_date(), 'dd-MM-yyyy')).display()
```
