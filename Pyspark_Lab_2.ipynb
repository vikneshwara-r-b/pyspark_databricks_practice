### Filter

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.format("csv") \
    .option("inferSchema", True) \
    .option("header", True) \
    .load("/Volumes/workspace/default/pyspark_lab_1_csv/BigMart Sales.csv")
```

```python
df.display()
```

### Scenario-1

```python
df1 = df.filter(col("Item_Fat_Content") == 'Low Fat').display()
```

### Scenario-2: AND condition
When having multiple conditions - each condition should be in curve brackets

```python
df2 = df.filter((col("Item_Fat_Content") == "Regular") & (col("Item_Weight") > 5.0))
```

```python
df2.display()
```

### Scenario-3: OR condition: isin()
When having multiple conditions - each condition should be in curve brackets

```python
df.filter(col("Item_Type").isin('Dairy','Meat','Household')).display()
```

### Scenario 4 - isNull()

```python
df.filter(col("Item_Weight").isNull()).display()
```

### With Column
- WithColumnRenamed - If you want to select all columns but rename only few of them
- Select + Alias - If you want to select only few columns, and rename some of them

#### Scenario-1: How to rename columns in DF

```python
df.select(col("Item_Fat_Content").alias("I_F_C"),
                col("Item_Type").alias("I_T")).display()
```

```python
df6 = df.withColumnRenamed("Item_Fat_Content", "I_F_C")
```

```python
df6.display()
```

```python
df.withColumn("New_Weight", lit("heavy")).display()
```

```python
df.withColumn("Product", col("Item_MRP")*col("Item_Weight")).display()
```

#### Scenario-2: Reg Exp Replace

This will be used when you have to replace any original value with any modified value from the parent dataframe

```python
df.withColumn("Item_Fat_Content", regexp_replace("Item_Fat_Content", "Regular", "Regular Fat")).display()
```

### TypeCasting
Used when you are converting data types of a column not from the schema but in the intermediate code

```python
df.withColumn("Outlet_Establishment_Year", col("Outlet_Establishment_Year").cast(StringType())).display()
```

### Order By and Sort By

```python
df.orderBy(col("Item_MRP").asc()).display()
```

```python
df.sort(col("Item_MRP").asc()).display()
```

```python
df.sort(col("Item_MRP").asc(), col("Item_Visibility").desc()).display()
```

```python
df_list_sort = df.sort(["Item_MRP", "Item_Visibility"], ascending=[1, 0])
```

```python
df_list_sort.display()
```

### groupByKey vs reduceByKey
- groupbykey - memory intensive operation, as driver is performing the aggregation work from output provided all executors
- reducebykey - Local Aggregation work is performed inside executor from output provided all executors
